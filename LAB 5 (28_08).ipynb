```python
import pandas as pd
import numpy as np
from sklearn.feature_selection import mutual_info_classif
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
  # Generate synthetic data
np.random.seed(42)
X = pd.DataFrame({
    'feature_1': np.random.rand(100) * 10,
    'feature_2': np.random.rand(100) * 5,
    'feature_3': np.random.randint(0, 3, 100) # Categorical-like feature
})
y = (X['feature_1'] + X['feature_2'] * 2 + np.random.randn(100) * 3 > 15).astype(int) # Binary target

print("Features (X) head:")
display(X.head())
print("\nTarget (y) head:")
display(y.head())
```

```python
# Calculate mutual information
mi_scores = mutual_info_classif(X, y)
mi_series = pd.Series(mi_scores, index=X.columns)

print("Mutual Information Scores:")
display(mi_series.sort_values(ascending=False))
```

```python
# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print(f"Training set size: {len(X_train)}")
print(f"Testing set size: {len(X_test)}")
```

```python
# Initialize and train the KNN classifier
# We'll choose k=5 neighbors for this example
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

# Make predictions on the test set
y_pred = knn.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"KNN Classifier Accuracy: {accuracy:.2f}")
```

```python
import pandas as pd
```

```python
students=["A", "B", "C", "D", "E", "F"]
study_hours = [2,4,6,8,10,12]
marks=[50,60,80,70,90,100]
```

```python
df=pd.DataFrame({"students":students,
                 "study_hours":study_hours, "marks":marks})
print("original data")
print(df)
```

```python
df["Rank X"] = df["study_hours"].rank(method="average",ascending=True).astype(int)
print(df)
```

```python
df["RANK Y"] = df["marks"].rank(method="average",ascending=True).astype(int)
print(df)
```

```python
df["d"] = df["Rank X"] - df["RANK Y"]
print(df)
```

```python
df["d^2"] = df["d"]**2
print(df)
```

```python
sum_d2 = df["d^2"].sum()
print("\nSum of d^2 = ",sum_d2)
n= len(df)
print("number of students (n)=",n)
```

```python
rho=1 - (6 * sum_d2)/(n * (n**2 - 1))
print(f"spearman's Rank Correlation coefficient (p)= {rho:.4f}")
```

```python
if rho > 0.9:
  print("\n interpretation: very strong posoitive correlation")
elif rho > 0.7:
 print("\n interpretation:  strong posoitive correlation")
elif rho > 0.3:
 print("\n interpretation: modarate posoitive correlation")
elif rho > 0:
  print("\n interpretation: weak posoitive correlation")
elif rho == 0:
  print("\n interpretation:No correlation")
else:
  print("\ninterpretation: Negative correlation")
```

```python
import numpy as np
import pandas as pd
import math
```

```python
data= np.array([
    [8,2],
    [5,3],
    [1,1]
])
rows=("High", "Medium", "Low")
columns=["Pass", "Fail"]
df=pd.DataFrame(data, index=rows, columns=columns)
print("Original Data")
print(df)
```

```python
total= data.sum()
print("\n total observation=", total)
```

```python
row_totals=data.sum(axis=1)
column_totals=data.sum(axis=0)
print("\n row_totals=", row_totals)
print("\n column_totals=", column_totals)
```

```python
print("MARGINAL PROBABILITIES")

print("\nP(Result):")
for i in range(len(rows)):
  probability=row_totals[i]/total
  print(f"P({rows[i]})={row_totals[i]/total}")

print("\nP(Grade):")
for j in range(len(columns)):
  print(f"P({columns[j]})={column_totals[j]/total}")
```

```python
MI=0
for i in range(len(rows)):
  for j in range(len(columns)):
    p_xy=data[i][j]/total
    p_x=row_totals[i]/total
    p_y=column_totals[j]/total
    ratio=p_xy/(p_x*p_y)
    contribution=p_xy*math.log2(ratio)
    MI=MI+contribution
    print(f"P({rows[i]},{columns[j]})={p_xy}")
    print(f"P({rows[i]})={p_x}")
    print(f"P({columns[j]})={p_y}")
    print(f"ratio={ratio}")
    print(f"contribution={contribution}")
    print(f"MI={MI}")
    print("----------------")
    print()
print(f"Mutual Information (MI)={MI}")
```
