```python
import base64
from google.colab import userdata
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage

with open("blood_work.png","rb") as f:
  image_b64 = base64.b64encode(f.read()).decode()

#Current Groq vision-capable model (check console.groq.com/docs/vision for the latest)
llm = ChatGroq(
    model="qwen/qwen3.8-27b",
    api_key=userdata.get("GROQ_API_KEY"),
)
message = HumanMessage(
    content=[
        {"type":"text","text":"This is a blood work report. Extract all test results and flag any values outside the normal range"},
        {"type":"image_url",
         "image_url":{"url":f"data:image/png;base64,{image_b64}"}}
    ]
)
response = llm.invoke([message])
print(response.content)
```

```python
from PIL import Image
import pytesseract

#Open the image file
image = Image.open("blood_work.png")

#Use Tesseact to do OCR on the image
extracted_text = pytesseract.image_to_string(image)
print("Exttracted text:\n", extracted_text)
message = HumanMessage(content=f"This is a blood work report. Extract all test results and flag any values outside the normal range. here is the report text:{extracted_text}")
```
